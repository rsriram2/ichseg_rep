```python
import pandas as pd
import numpy as np
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter
from reportlab.lib.utils import ImageReader
from PIL import Image
import os
```

```python
methods = ['v1', 'robust', 'synthstrip', 'hdctbet', 'ctbet', 'brainchop']

for method in methods:
    original_csv = pd.read_csv(f"/Users/rushil/ichseg/{method}/annotations.csv")
    failures_csv = pd.read_csv(f"/Users/rushil/ichseg/{method}/annotations_failures.csv")
    
    mask_original = (original_csv['1 - Neck'] == 'yes') & (original_csv['4 - Nonbrain'] == 'yes')
    mask_failures = (failures_csv['1 - Neck'] == 'yes') & (failures_csv['4 - Nonbrain'] == 'yes')
    
    filenames_original = original_csv.loc[mask_original, 'Filename'].tolist()
    filenames_failures = failures_csv.loc[mask_failures, 'Filename'].tolist()
    combined_filenames = filenames_original + filenames_failures
    unique_filenames = pd.unique(combined_filenames).tolist()
    
    path = f"/Users/rushil/ichseg/{method}/image_ss_{method}/"
    full_paths = [os.path.join(path, fname) for fname in unique_filenames]
    out_file = f"/Users/rushil/ichseg/{method}/check_neck.pdf"
    
    c = canvas.Canvas(out_file, pagesize=letter)
    width, height = letter

    for img_path in full_paths:
        if os.path.exists(img_path):
            try:
                img = Image.open(img_path)
                img_reader = ImageReader(img)
                c.drawImage(img_reader, 0, 0, width=width, height=height, preserveAspectRatio=True)
                c.showPage()
            except Exception as e:
                print(f"Could not add {img_path}: {e}")
        else:
            print(f"File not found: {img_path}")

    c.save()
    print(f"Saved PDF for {method} to {out_file}")
```
