```python
import pandas as pd
import numpy as np
```

```python
new_df = pd.DataFrame(pd.read_csv('/Users/rushil/ichseg/robust/image_ss_robust_more.csv'))
old_df = pd.DataFrame(pd.read_csv('/Users/rushil/ichseg/robust/annotations.csv'))
#find rows that have any 'yes' in any column
yes_rows = new_df[new_df.apply(lambda x: x.str.contains('yes', case=False, na=False).any(), axis=1)]
#add the yes_rows to the old_df
old_df = pd.concat([old_df, yes_rows], ignore_index=True)
old_df.drop_duplicates(subset='Filename', keep='first', inplace=True)

old_df.to_csv('/Users/rushil/ichseg/robust/annotations.csv', index=False)
```

```python
real_df = pd.DataFrame(pd.read_csv('/Users/rushil/ichseg/robust/annotations.csv'))
failures_df = pd.DataFrame(pd.read_csv('/Users/rushil/ichseg/robust/annotations_failures.csv'))
compare_df = pd.DataFrame(pd.read_csv('/Users/rushil/ichseg/synthstrip/annotations.csv'))

real_df['Filename'] = real_df['Filename'].str.strip()
compare_df['Filename'] = compare_df['Filename'].str.strip()
failures_df['Filename'] = failures_df['Filename'].str.strip()


#find the files that are in compare_df but not in real_df
missing_files_1 = compare_df[~compare_df['Filename'].isin(real_df['Filename'])]
missing_files_2 = compare_df[~compare_df['Filename'].isin(failures_df['Filename'])]


#add the Filename column from missing files to the Filename column in real_df
real_df = pd.concat([real_df, missing_files_1[['Filename']]], ignore_index=True)
failures_df = pd.concat([failures_df, missing_files_2[['Filename']]], ignore_index=True)

#for the values that are nans in real_df, fill them with 'no'
real_df.fillna('no', inplace=True)
failures_df.fillna('no', inplace=True)

real_df.to_csv('/Users/rushil/ichseg/robust/annotations_check.csv', index=False)
failures_df.to_csv('/Users/rushil/ichseg/robust/annotations_failures_check.csv', index=False)



```
